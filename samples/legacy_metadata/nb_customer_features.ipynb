# Customer feature notebook
Legacy Synapse notebook sample for migration analysis.

```python
customers = spark.sql("""
SELECT customer_id, customer_name, customer_segment
FROM legacy_lake.ext.ExternalCustomers
""")
```

```python
orders = spark.sql("""
SELECT customer_id, COUNT(*) AS order_count, SUM(order_total) AS lifetime_value
FROM legacy_lake.ext.ExternalOrders
GROUP BY customer_id
""")
```

```python
features = customers.join(orders, 'customer_id', 'left')
features.write.mode('overwrite').saveAsTable('sales.mart.DimCustomer')
```
